# Error Analysis
After training a machine learning model, it is essential to analyze its errors in order to understand its behavior and limitations

Error analysis helps identify:
- False positives (detection where no lesion exists)
- false negatives (lesions that the model fails to detect)
- Difficult images that challenge the model

Understanding these cases allows researchers to imporve datasets, refine annotations, and guide feature model improvements.

In this notebook, we analyze the predictions produced by the trained YOLO model on the test dataset.

## Import required libraries

```python
from ultralytics import YOLO
from pathlib import Path
import random
import cv2
import matplotlib.pyplot as plt
```

## Define project paths
We define the paths for the trained model and the test dataset

```python
PROJECT_ROOT = Path("..").resolve()

MODEL_PATH = PROJECT_ROOT / "ml" / "models" / "best.pt"

TEST_IMAGES_DIR = PROJECT_ROOT / "dataset" / "selected" / "images" / "test"

MODEL_PATH, TEST_IMAGES_DIR
```

## Load trained YOLO model

```python
model = YOLO(MODEL_PATH)
```

## Run predictions on the test dataset
We perform inference on the test images in order to inspect the predictions produced by the model. 

```python
results = model.predict(
    source=str(TEST_IMAGES_DIR),
    conf=0.25,
    save=False
)
```

## Function to visualize model predictions
This function overlays the predicted bounding boxes on the original image.

```python
def visualize_prediction(image_path):
    result = model(image_path)
    
    plotted = result[0].plot()
    
    plt.figure(figsize=(6,6))
    plt.imshow(plotted)
    plt.axis("off")
    plt.show()
```

## Visualize inspection of predictions
We randomly select images from the test dataset and visualize the model prediction.

```python
test_images = list(TEST_IMAGES_DIR.glob("*.*"))

sample_image = random.choice(test_images)

visualize_prediction(sample_image)
```

## Inspect multiple predictions

```python
for _ in range(5):
    img = random.choice(test_images)
    
    visualize_prediction(img)
```

## Observations
During visual inspection, we evaluate the following aspects:

- Whether bounding boxes correctly surround carious lessons
- Whether the model detects lesions that are not present
- Whether existing lesions are missed
- Whether lighting conditions or tooth overlap affects detection

These observations help identify weaknesses in the model and guide further improvements.

## Types of errors
Commom error types in object detection systems include:

False Positives
The model detects a lesion where none exists.

False Negatives
The model fails to detect an existing lesion

Localization Erros
The bounding box is incorrectly placed or poorly aligned with the lesion.

Understanding these errors provides insight into the limitations of the detection system.

## Conclusion

Error analysis is an essential step in machine learning research.

By visually inspecting model predictions, we can better understand the behavior of the detection system and identify potential improvements in both the dataset and the training procedure.
