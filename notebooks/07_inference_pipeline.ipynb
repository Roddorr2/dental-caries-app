# Inference Pipeline

After training and evaluating the object detection model, the next step is to build a simple inference pipeline.

The inference pipeline represents how the trained model will be used in practice. In this stage, the model receives an input image and returns predictions indicating the location of potential dental caries.

This notebook demonstrates the basic workflow for running inference using the trained YOLO model.

The pipeline consists of the following steps:

1. Load the trained model
2. Load an input image
3. Run model inference
4. Visualize the detected bounding boxes

This stage simulates how the model could be integrated into a clinical decision support system.


## Import required libraries

```python
from ultralytics import YOLO
from pathlib import Path
import matplotlib.pyplot as plt
```

## Define project paths

We define the path to the trained model and the dataset images that will be used for testing the inference pipeline.

```python
PROJECT_ROOT = Path("..").resolve()

MODEL_PATH = PROJECT_ROOT / "ml" / "models" / "best.pt"

TEST_IMAGES_DIR = PROJECT_ROOT / "dataset" / "selected" / "images" / "test"

MODEL_PATH, TEST_IMAGES_DIR
```

## Load trained YOLO model

```python
model = YOLO(MODEL_PATH)
```

## Select a sample image
We select a sample image from the test dataset to run inference

```python
sample_images = list(TEST_IMAGES_DIR.glob("*.*"))

sample_image = sample_images[0]

sample_image
```

## Run model inference

```python
results = model(sample_image)
```

## Visualize prediction results

The model output includes bounding boxes indicating the predicted location of dental caries.

```python
prediction_image = results[0].plot()

plt.figure(figsize=(6,6))
plt.imshow(prediction_image)
plt.axis("off")
```

## Create reusable inference function
To simplify the process, we create a function that runs the interence pipeline on any input image.

```python
def detect_caries(image_path) -> None:
    results = model(image_path)
    
    predicted = results[0].plot()
    
    plt.figure(figsize=(6,6))
    plt.imshow(predicted)
    plt.axis('off')
    plt.show()
```

## Test the inference function

```python
detect_caries(sample_image)
```

## Interpretation of results

The bounding boxes produced by the model indicate regions where dental caries are predicted to exist.

Each detection is associated with a confidence score representing the model's certainty.

This pipeline demonstrates the practical application of the trained model and serves as a prototype for integration into larger systems such as web-based diagnostic tools or clinical decision support system.
