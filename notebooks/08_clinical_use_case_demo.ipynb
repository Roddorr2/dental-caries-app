# Clinical Use Case Demonstration

This notebook demonstrates a simulated clinical workflow using the trained dental caries detection model.

The purpose of this notebook is not to train or evaluate the model, but to illustrate how the system could be used in a real-world dental environment.

In a potential clinical setting, the workflow could be as follows:

1. A dentist captures an intraoral image.
2. The image is uploaded into a diagnostic support system.
3. The trained model analyzes the image.
4. The system highlights possible carious lesions.
5. The dentist uses this information as an aid for diagnosis.

This notebook demonstrates a simplified version of that process using the trained YOLO model.

## Import required libraries

```python
from ultralytics import YOLO
from pathlib import Path
import matplotlib.pyplot as plt
```

## Define project paths

```python
PROJECT_ROOT = Path("..").resolve()

MODEL_PATH = PROJECT_ROOT / "ml" / "models" / "best.pt"

TEST_IMAGES_DIR = PROJECT_ROOT / "dataset" / "selected" / "images" / "test"

MODEL_PATH, TEST_IMAGES_DIR
```

## Load trained detection model

```python
model = YOLO(MODEL_PATH)
```

## Simulated clinical case
For demonstration purposes, we select an image from the dataset to simulate a patient case.
In a real system, this image would come from a dental camera or intraoral scanner.

```python
sample_images = list(TEST_IMAGES_DIR.glob("*.*"))

patient_case = sample_images[1]

patient_case
```

## Run diagnostic inference

```python
results = model(patient_case)
```

## Visualize diagnostic output

The system highlights regions where the model predicts the presence of dental caries.

```python
diagnostic_image = results[0].plot()

plt.figure(figsize=(6,6))
plt.imshow(diagnostic_image)
plt.axis("off")
```

## Detection summary
We can also extract structured from the model predictions.

```python
detections = results[0].boxes

num_detections = len(detections)

print(f"Detected lesions: {num_detections}")
```

## Interpretation in a clinical context

The highlighted bounding boxes represent areas where the model predicts potential dental caries.

It is important to emphasize that this system is intended as a **clinical decision support tool**, not a replacement for professional diagnosis.

The final diagnostic decision must always be made by a qualified dental professional.

Such systems could help dentists by:

- Highlighting suspicious regions
- Reducing oversight errors
- Supporting early detection of lesions
