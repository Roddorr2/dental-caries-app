# Model Evaluation

After training an object detection model, it is necessary to evaluate its performance using standardized metrics.
In object detection tasks, several metrics are commonly used to assess how well a model detects and localizes objects.
This notebook evaluates the trained YOLO model using the test dataset and reports the following metrics:

- Precision
- Recall
- Mean Average Precision at IoU (mAP50)
- Mean Average Precision across IoU thresholds (mAP50-95)

These metrics provide a quantitative assessment of the detection performance of the training model.

## Import required libraries

```python
from ultralytics import YOLO
from pathlib import Path
import pandas as pd
```

## Define project paths

We define the project root directory, the trained model path, and the dataset configuration file

```python
PROJECT_ROOT = Path("..").resolve()

MODEL_PATH = PROJECT_ROOT / "ml" / "models" / "best.pt"

DATA_YAML = PROJECT_ROOT / "ml" / "configs" / "data.yaml"

MODEL_PATH, DATA_YAML
```

## Load trained model

```python
model = YOLO(MODEL_PATH)
```

## Run model evaluation
We evaluate the trained model on the dataset defined in `data.yaml`.

The YOLO framework automatically calculates standard object detection metrics.

```python
metrics = model.val(data=str(DATA_YAML), split="test")
```

## Extract evaluation metrics

```python
precision = metrics.box.mp
recall = metrics.box.mr
map50 = metrics.box.map50
map50_95 = metrics.box.map

precision, recall, map50, map50_95
```

## Create evaluation summary table

```python
evaluation_results = pd.DataFrame({
    "Metric": ["Precision", "Recall", "mAP50", "mAP50-95"],
    "Value": [precision, recall, map50, map50_95]
})

evaluation_results
```

## Interpretation of metrics

Precision measures the proportion of predicted detections that are correct.

Recall measures the proportion of actual lesions that were successfully detected.

mAP50 represents the mean average precision at an intersection-over-union (IoU) threshold of 0.5.

mAP50–95 is a stricter metric that averages detection performance across multiple IoU thresholds.

Higher values indicate better detection performance.

## Conclusion

The evaluation metrics obtained in this notebook provide a quantitative assessment of the model’s ability to detect dental caries.

These results can be used to compare different model configurations, training strategies, or dataset variations in future experiments.
